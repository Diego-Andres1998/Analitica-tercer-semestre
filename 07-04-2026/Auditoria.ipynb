```python
import pandas as pd

# 1. Cargar el dataset
df = pd.read_csv('netflix1.csv', sep=';')
```

```python
print(df.info())
```

```python
# Conteo exacto de nulos por columna
print(df.isnull().sum())

# Ver filas duplicadas (totales)
print(f"Duplicados totales: {df.duplicated().sum()}")
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
# Elimina cualquier fila que tenga al menos un nulo
df.dropna(inplace=True)
```

```python
df.dropna(subset=['duration'], inplace=True)
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip())
```

```python
# Ver si hay títulos repetidos con diferentes IDs
print(df.title.nunique()) 
print(df.shape[0])
```

```python
# Conteo exacto de nulos por columna
print(df.isnull().sum())

# Ver filas duplicadas (totales)
print(f"Duplicados totales: {df.duplicated().sum()}")
```

```python
df.to_csv('dataset_limpio.csv', index=False, encoding='utf-8-sig')
```
